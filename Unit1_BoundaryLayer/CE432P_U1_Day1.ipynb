### 1. Theoretical Background
When a real fluid of kinematic viscosity $\nu$ flows past a solid boundary (e.g. flat plate, spillway chute, pier), viscous effects are concentrated within a thin layer adjacent to the surface called the **Boundary Layer** ($\delta$).

#### Key Definitions:
1. **Nominal Boundary Layer Thickness ($\delta$):** Distance from the boundary where velocity $u = 0.99 U_\infty$.
2. **Displacement Thickness ($\delta^*$):** Distance by which the solid boundary would have to be displaced to account for the reduction in mass flow rate:
$$\delta^* = \int_0^\delta \left(1 - \frac{u}{U_\infty}\right) dy$$
3. **Momentum Thickness ($	heta$):** Loss of momentum flux represented as an equivalent layer:
$$	heta = \int_0^\delta \frac{u}{U_\infty}\left(1 - \frac{u}{U_\infty}\right) dy$$
4. **Energy Thickness ($\delta^{**}$):** Loss of kinetic energy flux:
$$\delta^{**} = \int_0^\delta \frac{u}{U_\infty}\left(1 - \frac{u^2}{U_\infty^2}\right) dy$$
5. **Shape Factor ($H$):**
$$H = \frac{\delta^*}{	heta} \quad (H \approx 2.59 \text{ for laminar}, \, H \approx 1.3 \text{ to } 1.4 \text{ for turbulent})$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Comparison of Velocity Profiles:
# 1. Linear: u/U = eta
# 2. Parabolic: u/U = 2*eta - eta**2
# 3. Sinusoidal: u/U = sin(pi/2 * eta)
# 4. Turbulent 1/7th Power Law: u/U = eta**(1/7)

eta = np.linspace(0, 1, 200)

u_linear = eta
u_parabolic = 2 * eta - eta**2
u_sinusoidal = np.sin(np.pi / 2 * eta)
u_turbulent = eta**(1/7)

plt.figure(figsize=(8, 6))
plt.plot(u_linear, eta, label="Linear ($u/U = \eta$)", lw=2)
plt.plot(u_parabolic, eta, label="Parabolic ($2\eta - \eta^2$)", lw=2)
plt.plot(u_sinusoidal, eta, label="Sinusoidal ($\sin(\pi\eta/2)$)", lw=2)
plt.plot(u_turbulent, eta, label="Turbulent ($1/7^{th}$ Power)", lw=2.5, color="red")

plt.xlabel("Normalized Velocity ($u / U_\infty$)", fontsize=12)
plt.ylabel("Normalized Distance ($y / \delta$)", fontsize=12)
plt.title("Comparison of Boundary Layer Velocity Profiles", fontsize=14, fontweight="bold")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=11)
plt.xlim(0, 1.05)
plt.ylim(0, 1)
plt.show()
```

```python
# Compute Displacement & Momentum Thickness for Parabolic Profile
# u/U = 2*eta - eta^2
# delta* = delta * integral(1 - (2*eta - eta^2)) d_eta from 0 to 1
#        = delta * [eta - eta^2 + eta^3/3]_0^1 = delta * (1 - 1 + 1/3) = delta / 3

delta_val = 0.015 # 15 mm nominal boundary layer thickness
delta_star = delta_val * np.trapz(1 - u_parabolic, eta)
theta = delta_val * np.trapz(u_parabolic * (1 - u_parabolic), eta)
shape_factor = delta_star / theta

print(f"Nominal Boundary Layer Thickness (delta) : {delta_val * 1000:.2f} mm")
print(f"Displacement Thickness (delta*)          : {delta_star * 1000:.2f} mm (Ratio: {delta_star/delta_val:.3f})")
print(f"Momentum Thickness (theta)              : {theta * 1000:.2f} mm (Ratio: {theta/delta_val:.3f})")
print(f"Shape Factor (H = delta* / theta)       : {shape_factor:.3f}")
```
