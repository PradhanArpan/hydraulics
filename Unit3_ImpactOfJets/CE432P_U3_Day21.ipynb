### 1. Velocity Triangles & Euler Turbine Equation
When a jet of area $a$ and velocity $V_1$ strikes a curved vane moving with tangential velocity $u_1$:
1. **Relative Velocity at Inlet:**
$$\vec{V}_{r1} = \vec{V}_1 - \vec{u}_1$$
2. **Relative Velocity at Outlet:** If smooth with friction factor $k$, $V_{r2} = k V_{r1}$.
3. **Hydrodynamic Force in Direction of Motion:**
$$F_x = \rho a V_{r1} (V_{w1} \pm V_{w2})$$
4. **Work Done per Second ($P$):**
$$W = F_x \cdot u = \rho a V_{r1} (V_{w1} \pm V_{w2}) u$$
5. **Hydraulic Efficiency ($\eta_h$):**
$$\eta_h = \frac{2 u (V_{w1} \pm V_{w2})}{V_1^2}$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter sweep: Blade Speed Ratio (phi = u / V1) vs Efficiency
# For a Pelton bucket with deflection angle theta = 165 deg (outlet angle beta2 = 15 deg)
# eta_h = 2 * (u / V1) * (1 - u / V1) * (1 + k * cos(beta2))

phi = np.linspace(0.01, 0.99, 200)
k_friction = 0.95 # friction coefficient
beta2_deg = 15.0 # outlet angle
beta2_rad = np.radians(beta2_deg)

eta_h = 2 * phi * (1 - phi) * (1 + k_friction * np.cos(beta2_rad))

opt_idx = np.argmax(eta_h)
phi_opt = phi[opt_idx]
eta_opt = eta_h[opt_idx]

plt.figure(figsize=(8, 5.5))
plt.plot(phi, eta_h * 100, lw=2.5, color="darkred", label="Efficiency ($\eta_h$)")
plt.scatter([phi_opt], [eta_opt * 100], color="black", s=70, zorder=5, 
            label=f"Optimum: $\phi = {phi_opt:.2f}, \eta_{{max}} = {eta_opt*100:.1f}\%$")
plt.axvline(0.5, color="gray", linestyle="--", alpha=0.7, label="Theoretical $\phi = 0.5$")

plt.xlabel("Speed Ratio $\phi = u / V_1$", fontsize=12)
plt.ylabel("Hydraulic Efficiency $\eta_h$ (%)", fontsize=12)
plt.title("Pelton Vane Efficiency vs Speed Ratio", fontsize=14, fontweight="bold")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=11)
plt.xlim(0, 1)
plt.ylim(0, 100)
plt.show()
```
