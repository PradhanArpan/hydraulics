### 1. System Curve and Operating Point
The total head required by a pumping system consists of the static lift $H_{stat}$ plus friction and minor losses:
$$H_{sys} = H_{stat} + K Q^2$$

The centrifugal pump delivers a head approximated by the second-order pump characteristic:
$$H_{pump} = H_0 - A Q^2$$

The intersection $H_{sys} = H_{pump}$ establishes the **Operating Point** $(Q_{op}, H_{op})$.


```python
import numpy as np
import matplotlib.pyplot as plt

# Pump Performance Curve: H = 45 - 1200 * Q^2 (Q in m3/s, H in m)
Q = np.linspace(0, 0.16, 200) # discharge in m3/s
H_pump = 45.0 - 1200.0 * (Q**2)
eta_pump = 100 * (15 * Q - 55 * Q**2) # efficiency %

# System Head Curves for two static lifts
H_static1 = 18.0
H_static2 = 25.0
K_loss = 1600.0 # pipe resistance coefficient

H_sys1 = H_static1 + K_loss * (Q**2)
H_sys2 = H_static2 + K_loss * (Q**2)

# Solve operating points: H_0 - A*Q^2 = H_stat + K*Q^2 => Q^2 = (H_0 - H_stat) / (A + K)
Q_op1 = np.sqrt((45.0 - H_static1) / (1200.0 + K_loss))
H_op1 = H_static1 + K_loss * (Q_op1**2)

fig, ax1 = plt.subplots(figsize=(9, 6))

ax1.plot(Q * 1000, H_pump, lw=2.5, color="blue", label="Pump Head Curve $H_{pump}$")
ax1.plot(Q * 1000, H_sys1, lw=2, color="green", label=f"System 1 ($H_{{stat}}={H_static1}$m)")
ax1.plot(Q * 1000, H_sys2, lw=2, color="orange", linestyle="--", label=f"System 2 ($H_{{stat}}={H_static2}$m)")
ax1.scatter([Q_op1 * 1000], [H_op1], color="red", s=90, zorder=5, label=f"Duty Point 1: {Q_op1*1000:.1f} L/s @ {H_op1:.1f} m")

ax1.set_xlabel("Discharge $Q$ (Liters/second)", fontsize=12)
ax1.set_ylabel("Head $H$ (m)", fontsize=12, color="blue")
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.set_ylim(0, 50)
ax1.set_xlim(0, 160)

ax2 = ax1.twinx()
ax2.plot(Q * 1000, eta_pump, color="purple", linestyle=":", lw=2, label="Efficiency ($\eta$)")
ax2.set_ylabel("Efficiency (%)", fontsize=12, color="purple")
ax2.set_ylim(0, 100)

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right", fontsize=10.5)

plt.title("Centrifugal Pump Selection & System Matching", fontsize=14, fontweight="bold")
plt.show()
```
